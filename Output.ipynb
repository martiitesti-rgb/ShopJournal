```python
import json

selected_ids = {
    "ggf-note-000002",
    "ggf-note-000048",
    "ggf-note-000013",
    "ggf-note-000124",
    "ggf-note-000010",
    "ggf-note-000170",
    "ggf-note-000001",
    "ggf-note-000039",
    "ggf-note-000015",
    "ggf-note-000058",
    "ggf-note-000022",
    "ggf-note-000003",
    "ggf-note-000008",
    "ggf-note-000072",
    "ggf-note-000005",
}

selected_notes = []
with open("/Users/martitesti/Desktop/uni/tesi/ShopJournal/ggf_notes_student_v0.1.jsonl", "r") as f:
    for line in f:
        note = json.loads(line)
        if note["note_id"] in selected_ids:
            selected_notes.append(note)

print(f"Notes: {len(selected_notes)}")
for n in selected_notes:
    print(f"  {n['note_id']} — {n['note_text'][:60]}")
```

```python
#query+note
queries = [
    {"query_id": "q01", "note_id": "ggf-note-000002", "query": "I'm looking for a chocolate snack"},
    {"query_id": "q02", "note_id": "ggf-note-000048", "query": "I need a good dark roast coffee"},
    {"query_id": "q03", "note_id": "ggf-note-000013", "query": "I want to send a food gift to a friend"},
    {"query_id": "q04", "note_id": "ggf-note-000124", "query": "looking for an organic herbal tea"},
    {"query_id": "q05", "note_id": "ggf-note-000010", "query": "I need something for breakfast, quick"},
    {"query_id": "q06", "note_id": "ggf-note-000170", "query": "I want candy for a birthday party"},
    {"query_id": "q07", "note_id": "ggf-note-000001", "query": "I need bone broth or a soup base"},
    {"query_id": "q08", "note_id": "ggf-note-000039", "query": "looking for a vegan protein bar"},
    {"query_id": "q09", "note_id": "ggf-note-000015", "query": "I want a quick healthy frozen meal"},
    {"query_id": "q10", "note_id": "ggf-note-000058", "query": "I need canned goods for the pantry"},
    {"query_id": "q11", "note_id": "ggf-note-000022", "query": "looking for whole grain cookies or crackers"},
    {"query_id": "q12", "note_id": "ggf-note-000003", "query": "I want a healthy snack with dried fruit"},
    {"query_id": "q13", "note_id": "ggf-note-000008", "query": "I'm looking for a cherry or vegetable juice"},
    {"query_id": "q14", "note_id": "ggf-note-000072", "query": "I need a ready-to-eat chicken meal, low calorie"},
    {"query_id": "q15", "note_id": "ggf-note-000005", "query": "looking for a good sea salt or cooking spice"},
]
```

```python
import random

notes_dict = {n["note_id"]: n for n in selected_notes}

def build_pool(query_text, note, df, n_keyword=20, n_random=10):
    
    query_words = query_text.lower().split()
    note_terms = note["distinctive_terms"]
    # joining all the keywords
    all_keywords = list(set(query_words + note_terms))
    
   
     #selecting the first random 20 products
    mask = df["title"].str.lower().str.contains("|".join(all_keywords), na=False)
    keyword_matches = df[mask].head(n_keyword)
    
    # adding the negatives ( those products that do not contain the key word)
    random_sample = df[~mask].sample(n=n_random, random_state=42)
    
    # deleting duplicates
    pool = pd.concat([keyword_matches, random_sample]).drop_duplicates(subset=["asin"])
    
    return pool[["asin", "title", "price"]].to_dict(orient="records")


all_pools = []
for q in queries:
    note = notes_dict[q["note_id"]]
    pool = build_pool(q["query"], note, df)
    all_pools.append({
        "query_id": q["query_id"],
        "query": q["query"],
        "note_id": q["note_id"],
        "note_text": note["note_text"],
        "candidates": pool
    })
    print(f"{q['query_id']} — {len(pool)} candidates")

    
```

```python
with open("pool_candidates.jsonl", "w") as f:
    for item in all_pools:
        f.write(json.dumps(item) + "\n")

print("Pool saved in pool_candidates.jsonl")
```

```python
import csv

with open("labelling.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["query_id", "query", "note_text", "asin", "title", "price", "label", "incerto", "note_labelling"])
    
    for pool in all_pools:
        for c in pool["candidates"]:
            price = c["price"] if c["price"] not in (None, "None") else ""
            writer.writerow([
                pool["query_id"],
                pool["query"],
                pool["note_text"],
                c["asin"],
                c["title"],
                price,
                "",   # label 
                "",   # incerto 
                "",   # note_labelling 
            ])

print("File labelling.csv")
```
