This paper evaluates, through accuracy metrics, the performance of several collaborative filtering algorithms in pursuing the top-N recommendation task.
The key aspects are the two primary approaches to collaborative filtering, along with a standard baseline:
- Non-personalised models: These present to any user a predefined, fixed list of items, regardless of his/her preferences. As stated in the paper, this should be used as a baseline for the personalised algorithms, not as the final one.

- The neighborhood approach (commonly called Memory-based): This approach is based on the similarity among either users or items. Algorithms centered on user-user similarity predict the rating by a user based on the ratings expressed by users similar to him about such item.

- The latent factor approach (Model-based): A method based on factoring the user-item ratings matrix.

Recommender Systems Based on User Reviews: The State of the Art

In here, we have a general view of how reviews can improve the performance of standard recommending approaches (such as content-based recommending, collaborative filtering, and preference-based product ranking). Indeed, information obtained from reviews is useful as:

- It can help deal with the problem of large data sparsity by providing additional information about user preferences.

- It can help solve the cold-start problem for new users.

- It has been used to determine rating quality, to help derive users’ context-dependent preferences, and to learn users’ latent preference factors by considering the aspect opinions mentioned in reviews.

There are two main categories of review-based systems:

- Review-based user profile building, which emphasizes exploiting reviews to construct a user’s profile. This category can be further divided into three subcategories:

1. Term-based profile: Built by extracting frequent terms from reviews, which
differs from the traditional term profile that is built with the item’s static descriptions. The advantage is that the review terms can be more representative of the user’s personal characteristics, including their subjective preferences and writing style. Another advantage is that it enables the content-based recommending technique to be feasibly applied to a broader scope of products.

2. Rating profile: Uses reviews to either infer or enhance ratings.

3. Feature preference: Embodies users’ multifaceted opinions in reviews. Actually, a rating can only tell how much a user likes an item, whereas a feature preference can reveal why they like it. This preference can be directly expressed as an ⟨aspect, sentiment⟩ pair. The proven benefits of incorporating this information include not only the improvement of recommendations for users with review histories but also allowing the system to more effectively elicit a new user’s feature preferences.

Review-based product profile building, which focuses on using reviews to build informative product profiles.

Another topic covered is what type of information can be extracted from the reviews. I will summarize those that I think are the most valuable to us:

- Frequent terms: Can be used to characterize the reviewer with a term-based user profile.

- Overall opinions: A user’s sentiment orientation towards an item can be inferred from a review to represent their overall opinion.

- Feature opinions: Fine-grained opinions about specific features of an item can be extracted. Typical approaches include statistics-based methods, such as capturing frequently occurring nouns/phrases as feature candidates.

- Review emotions: Emotion reflects a reviewer’s mood (e.g., sadness, joy, distress, happiness) when writing the review, often indicated by words or emoticons.

- Comparative and contextual opinions.
