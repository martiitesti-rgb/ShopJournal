```python
!pip install datasets==2.20.0 pandas
from datasets import load_dataset
import pandas as pd
```

```python
### load of the entire dataset
dataset = load_dataset("McAuley-Lab/Amazon-Reviews-2023", "raw_review_All_Beauty", trust_remote_code=True)
print(dataset["full"][0]) #entire review of the first item of the dataset raw_review_All_Beauty
```

```python
#extracting the first 20 reviews
print("extracting the first 20 reviews:")
subset = dataset["full"].take(20)

#panda's dataframe
df_subset = pd.DataFrame(list(subset))

print("\n extracting title text and rating")
df_firstThree = df_subset[['title', 'text', 'rating']]
print(df_firstThree.head())
```
